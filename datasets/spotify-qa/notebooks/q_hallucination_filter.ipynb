```python
MODEL = "gpt-4o-mini-2024-07-18"
Q_EXTRACTION_NAME = "q_extraction_v0.1"
IDS_LIST_NAME = "initial_ids_v0.1"
MIN_MATCH_PERC = 90

IDS_OUTFILE = f"../output/lists/q_extraction/{MODEL}/{Q_EXTRACTION_NAME}/{IDS_LIST_NAME}_match_rate_{MIN_MATCH_PERC}.lst"
```

### Run

```python
import pandas as pd
import numpy as np

results_file = f"../output/q_extraction/parsed/{MODEL}/{Q_EXTRACTION_NAME}/{IDS_LIST_NAME}.tsv"

df = pd.read_csv(results_file, sep="\t")

# % of rows with match rate > values:
values = [.5, .75, .8, .9, .95]
pd.DataFrame({v: df[['match_rate_q']].gt(v).mean() for v in values})
```

```python
# Print some verbatim q to inspect them:
i = 0
for _, row in df.sort_values("match_rate_q", ascending=False).iterrows():
    print(f'[{row["match_rate_q"]:.2f}]', row["q_verbatim"], "||", row["question"])
    print(f'\t{row["reasoning"]}')
    i += 1
    if i > 50:
        break
```

```python
# Save IDs with both match rates > MIN_MATCH_PERC:
from pathlib import Path

min_match_frac = MIN_MATCH_PERC / 100
ids = df.query(f"match_rate_q > {min_match_frac}")["podcast_id"]

Path(IDS_OUTFILE).parent.mkdir(parents=True, exist_ok=True)

with open(IDS_OUTFILE, "w") as f:
    for id_ in ids:
        f.write(f"{id_}\n")
```

```python
print(f"{len(ids)}/{len(df)} questions")
print(f"Saved to: {IDS_OUTFILE}")
```
